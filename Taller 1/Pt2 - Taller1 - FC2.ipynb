**Ejercicio:**

2. Diseñe un programa en Fortran que calcule la altura máxima de un proyectil, conociendo el ángulo respecto a la horizontal y la rapidez inicial.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Escritura del código:

```python
%%writefile taller1_fc2.f90
program taller1_fc2
    implicit none

    ! Diccionario de datos: constante
    real, parameter :: g = 9.8         ! Constante de la gravedad proporcionada [m/s^2]
    real, parameter :: pi = 3.14159265 ! Constante pi (adimensional)

    ! Diccionario de datos: variables
    real :: angulo                     ! Ángulo proporcionado por el usuario [°]
    real :: angulo_rad                 ! Conversión del ángulo proporcionado por el usuario [rad] (adimensional)
    real :: v_0                        ! Rapidez inicial proporcionada por el usuario [m/s]
    real :: y_0                        ! Altura inicial proporcionada por el usuario [m]
    real :: y_max                      ! Altura máxima cálculada [m]


    ! Introducción al usuario:
    print *, "Hola usuario!"
    print *, "Por favor, introduzca el ángulo [°] (con respecto a la horizontal) con el cual será lanzado el proyectil:"
    read *, angulo
    print *, "Ahora, por favor introduzca la rapidez inicial [m/s] con la cual será lanzado el proyectil:"
    read *, v_0
    print *, "Por último, por favor introduzca la altura inicial [m] desde la cual será lanzado el proyectil:"
    read *, y_0

    ! Cálculos:
    angulo_rad = (angulo*2*pi) / 360                ! Conversión del ángulo a radianes

    y_max = y_0 + ( ((v_0**2)*(sin(angulo_rad)**2)) / (2*g) )
                                       ! Aplicación de la fórmula

    ! Reporte:
    print *, "Así, la altura máxima del proyectil será:", y_max, "m"

  end program taller1_fc2
```
